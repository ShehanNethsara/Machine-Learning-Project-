```python
# Check target variable distribution

attrition_counts = df["Attrition"].value_counts()

print(attrition_counts)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Visualize employee attrition distribution
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x="Attrition")

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Analyze attrition by overtime status
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="OverTime", hue="Attrition")

plt.title("Employee Attrition by Overtime Status")
plt.xlabel("OverTime")
plt.ylabel("Number of Employees")
plt.legend(title="Attrition")

plt.show()
```

```python
# Analyze attrition by job satisfaction
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="JobSatisfaction", hue="Attrition")

plt.title("Employee Attrition by Job Satisfaction")
plt.xlabel("Job Satisfaction Level")
plt.ylabel("Number of Employees")
plt.legend(title="Attrition")

plt.show()
```

```python
# Analyze attrition by job role
plt.figure(figsize=(12, 6))

sns.countplot(data=df, x="JobRole", hue="Attrition")

plt.title("Employee Attrition by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Attrition")

plt.tight_layout()
plt.show()
```

```python
# Analyze age distribution by attrition
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Age",
    hue="Attrition",
    multiple="stack",
    bins=15
)

plt.title("Age Distribution by Employee Attrition")
plt.xlabel("Age")
plt.ylabel("Number of Employees")

plt.show()
```
