```python
import matplotlib.pyplot as plt
import seaborn as sns

# Visualize employee attrition distribution
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x="Attrition")

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Analyze attrition by overtime status
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="OverTime", hue="Attrition")

plt.title("Employee Attrition by Overtime Status")
plt.xlabel("OverTime")
plt.ylabel("Number of Employees")
plt.legend(title="Attrition")

plt.show()
```

```python
# Analyze attrition by job satisfaction
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="JobSatisfaction", hue="Attrition")

plt.title("Employee Attrition by Job Satisfaction")
plt.xlabel("Job Satisfaction Level")
plt.ylabel("Number of Employees")
plt.legend(title="Attrition")

plt.show()
```

```python
# Analyze attrition by job role
plt.figure(figsize=(12, 6))

sns.countplot(data=df, x="JobRole", hue="Attrition")

plt.title("Employee Attrition by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Attrition")

plt.tight_layout()
plt.show()
```

```python
# Analyze age distribution by attrition
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Age",
    hue="Attrition",
    multiple="stack",
    bins=15
)

plt.title("Age Distribution by Employee Attrition")
plt.xlabel("Age")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Create Monthly Income groups for analysis
df["MonthlyIncomeGroup"] = pd.qcut(
    df["MonthlyIncome"], q=5,
    labels=["Very Low", "Low", "Medium", "High", "Very High"]
)

# Calculate attrition rate (%) by income group
income_attrition = pd.crosstab(
    df["MonthlyIncomeGroup"], df["Attrition"], normalize="index"
) * 100

income_attrition
```

```python
# Plot attrition rate by monthly income group

income_attrition["Yes"].plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Attrition Rate by Monthly Income Group")
plt.xlabel("Monthly Income Group")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.ylim(0, 50)

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
# Calculate attrition rate (%) by Years at Company
years_attrition = pd.crosstab(
    df["YearsAtCompany"], df["Attrition"], normalize="index"
) * 100

years_attrition
```

```python
# Plot attrition rate by years at company group

years_attrition["Yes"].plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Attrition Rate by Years at Company")
plt.xlabel("Years at Company")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
# Drop constant / irrelevant columns
constant_cols = ["EmployeeCount", "StandardHours", "Over18"]
df = df.drop(columns=[c for c in constant_cols if c in df.columns])
df = df.drop(columns=["EmployeeNumber"])  # ID column, not predictive

print("Shape after dropping columns:", df.shape)
```

```python
# Encode target variable (Attrition: Yes -> 1, No -> 0)
df["Attrition"] = df["Attrition"].map({"Yes": 1, "No": 0})

df["Attrition"].value_counts()
```

```python
# Create new engineered features
df["TenurePerCompany"] = df["YearsAtCompany"] / (df["NumCompaniesWorked"] + 1)
df["IncomePerYearAtCompany"] = df["MonthlyIncome"] / (df["YearsAtCompany"] + 1)
df["PromotionGap"] = df["YearsAtCompany"] - df["YearsSinceLastPromotion"]

df[["TenurePerCompany", "IncomePerYearAtCompany", "PromotionGap"]].head()
```

```python
# Bin Age into groups
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 45, 60],
    labels=["18-25", "26-35", "36-45", "46-60"]
)

df["AgeGroup"].value_counts()
```

```python
# Bin MonthlyIncome into groups
df["IncomeGroup"] = pd.qcut(
    df["MonthlyIncome"], q=4,
    labels=["Low", "Medium", "High", "Very High"]
)

df["IncomeGroup"].value_counts()
```
